```python
import statistics as st

num = [10,20, 30, 40, 1000]

media = st.mean(num)
mediana = st.median(num)
moda = st.mode(num)

print(f"A média é: {media}")
print(f'A mediana é: {mediana}')
print(f'A moda é: {moda}')
```

```python
ven_vend = {'Diogo' : 25, 'Gabriel' : 10, 'Razo' : 50}
ven_vend.update({'Enzo': 30, 'Caio': 50})
med_ven = st.mean(ven_vend.values())
nome_vend = list(ven_vend.keys())
print(f'A média de vendas é: {med_ven}')
print(nome_vend)
print(ven_vend)
print('\n')
print(f"{'Vendedor':<12} | {'Vendas':>6} | {'Média':>6} | {'Status':<12}")
print("-" * 45)

for vendedor, vendas in ven_vend.items():
    status = "Acima" if vendas >= med_ven else "Abaixo"
    print(f"{vendedor:<12} | {vendas:>6} | {med_ven:>6.1f} | {status:<12}")

# o < joga para esquerda, o > joga para a direita; o numero indica os caracteres para alinhar o organizar a tabela, se colocar um número menor ele preenche o valor a força
```
